```python
import os
from pathlib import Path

from dotenv import load_dotenv
from aieng.forecasting.data.adapters.fred import FREDAdapter


def find_repo_root() -> Path:
    """Find the repository root from the notebook's current working directory."""
    current = Path.cwd().resolve()
    for candidate in (current, *current.parents):
        if (candidate / "aieng-forecasting").is_dir() and (candidate / "pyproject.toml").exists():
            return candidate
    raise RuntimeError("Could not find the agentic-forecasting repository root.")


ROOT = find_repo_root()
load_dotenv(ROOT / ".env", override=False)

fred_api_key = os.getenv("FRED_API_KEY")
if not fred_api_key:
    raise RuntimeError(
        f"FRED_API_KEY is missing. Add it to {ROOT / '.env'} before running this cell."
    )

cad_usd_rate = "DEXCAUS"
oil_barrel = "DCOILWTICO"

cad_adapter = FREDAdapter(series_id=cad_usd_rate, api_key=fred_api_key)
oil_adapter = FREDAdapter(series_id=oil_barrel, api_key=fred_api_key)

cad_usd_df = cad_adapter.fetch()
oil_df = oil_adapter.fetch()

print("CAD/USD observations:")
print(cad_usd_df.head(15))
print("\nWTI observations:")
print(oil_df.head(15))
```
